```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()

for x in ["workclass", "education", "marital-status", "occupation", "relationship", "race", "gender", "native-country", "income"] :
  df1[x]=le.fit_transform(df1[x])
  print(df1[x])
```

```python
for x in ["age", "fnlwgt", "educational-num", "capital-gain", "capital-loss", "hours-per-week"] :
  Q1=df1[x].quantile(0.25)
  Q3=df1[x].quantile(0.75)

  IQR=Q3-Q1

  lowerbound=Q1- 1.5*IQR
  upperbound=Q3+ 1.5*IQR

  df1=df1[(df1[x] >= lowerbound) & (df1[x] <= upperbound)]

  print(df1.shape)
```
